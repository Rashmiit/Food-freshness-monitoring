```python
import os

# Path to your dataset
dataset_path = '/content/drive/MyDrive/Apple'

# List files in the dataset directory
image_files = os.listdir(dataset_path)
print("Files in dataset directory:", image_files)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

# Load and display a single image
def load_and_display_single_image(file_name):
    image_path = os.path.join(dataset_path, file_name)
    try:
        image = Image.open(image_path)
        plt.imshow(image)
        plt.axis('off')
        plt.show()
    except Exception as e:
        print(f"Error loading image {file_name}: {e}")

# Load and display the first image found
if len(image_files) > 0:
    load_and_display_single_image(image_files[0])
else:
    print("No image files found.")
```

```python
from PIL import Image
import numpy as np

# Function to load images and labels
def load_images(dataset_path, target_size=(128, 128)):
    images = []
    labels = []
    for file in os.listdir(dataset_path):
        if file.endswith(('.png', '.jpg', '.jpeg')):
            try:
                image = Image.open(os.path.join(dataset_path, file))
                image = image.resize(target_size)
                image = np.array(image)
                images.append(image)
                # Extract label from filename, adjust if necessary
                label = file.split('_')[1].replace('day', '')
                labels.append(int(label))
            except Exception as e:
                print(f"Error loading image {file}: {e}")
    return np.array(images), np.array(labels)

# Load images
images, labels = load_images(dataset_path)
print("Number of images loaded:", len(images))
```

```python
import re
from PIL import Image
import numpy as np
import os  # Import os for listing files

# Function to load images and labels
def load_images(dataset_path, target_size=(128, 128)):
    images = []
    labels = []
    for file in os.listdir(dataset_path):
        if file.endswith(('.png', '.jpg', '.jpeg')):
            try:
                # Open and resize the image
                image = Image.open(os.path.join(dataset_path, file))
                image = image.resize(target_size)
                image = np.array(image)
                images.append(image)

                # Extract day from filename using regular expressions
                match = re.search(r'day(\d+)', file)  # Extract 'day1', 'day2', etc.
                if match:
                    label = int(match.group(1))  # Convert '1', '2', etc. to integer
                else:
                    label = -1  # Assign a default value if no day is found

                labels.append(label)
            except Exception as e:
                print(f"Error loading image {file}: {e}")

    # Convert lists to numpy arrays
    images = np.array(images)
    labels = np.array(labels)

    # Normalize the images
    images = images / 255.0

    return images, labels

# Path to your dataset
dataset_path = '/content/drive/MyDrive/Apple'

# Load images and labels
images, labels = load_images(dataset_path)
print("Number of images loaded:", len(images))
print("Labels:", labels)
```

```python
import matplotlib.pyplot as plt

# Function to display images
def display_images(images, labels, num_images=10):
    num_images = min(num_images, len(images))
    plt.figure(figsize=(10, 10))
    for i in range(num_images):
        plt.subplot(1, num_images, i + 1)
        plt.imshow(images[i])  # Show image
        plt.title(f"Label: {labels[i]}")
        plt.axis('off')
    plt.show()

# Display a few sample images
display_images(images, labels)
```

```python
import matplotlib.pyplot as plt

# Function to display all images with labels
def display_all_images(images, labels, num_images_per_row=5):
    num_images = len(images)
    num_rows = (num_images + num_images_per_row - 1) // num_images_per_row
    plt.figure(figsize=(15, 3 * num_rows))

    for i in range(num_images):
        plt.subplot(num_rows, num_images_per_row, i + 1)
        plt.imshow(images[i])
        plt.title(f"Label: {labels[i]}")
        plt.axis('off')

    plt.tight_layout()
    plt.show()

# Display all images
display_all_images(images, labels)
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models
from sklearn.model_selection import train_test_split
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(images, labels, test_size=0.2, random_state=42)
```

```python
# Building the CNN model
model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(128, 128, 3)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(1)  # Output layer for regression
])
```

```python
# Compile the model
model.compile(optimizer='adam', loss='mean_squared_error', metrics=['mae'])
```

```python
# Train the model
history = model.fit(X_train, y_train, epochs=20, validation_split=0.2)
```

```python
# Evaluate the model on the test data
loss, mae = model.evaluate(X_test, y_test)
print(f"Mean Absolute Error on test data: {mae}")
```

```python
# Function to predict shelf life
def predict_shelf_life(image_path):
    image = Image.open(image_path)
    image = image.resize((128, 128))
    image = np.array(image) / 255.0
    image = np.expand_dims(image, axis=0)  # Add batch dimension
    predicted_day = model.predict(image)
    return predicted_day[0][0]
```

```python
# Example prediction
shelf_life = predict_shelf_life('/content/drive/MyDrive/Apple/apple_day17 (3).jpg')
print(f"Predicted shelf life (days): {shelf_life:.2f}")
```
